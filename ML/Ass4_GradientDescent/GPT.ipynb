```python
# Implement Gradient Descent Algorithm to find the local minima of a function.
# For example, find the local minima of the function y=(x+3)² starting from the point
# x=2


import matplotlib.pyplot as plt

def f(x):
    return (x+3)**2

def derivative(x):
    return 2*(x+3)

x = 2
learning_rate = 0.1
iterations = 50

points = [x]

for i in range(iterations):
    x = x - learning_rate * derivative(x)
    points.append(x)


x_values = range(-6, 4)
y_values = [f(i) for i in x_values]

plt.plot(x_values, y_values)
plt.scatter(points, [f(i) for i in points], color="red")
plt.scatter(points[0], f(points[0]), color="green")
plt.scatter(points[-1], f(points[-1]), color="blue")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Gradient Descent Function")
plt.grid()
plt.show()
```
