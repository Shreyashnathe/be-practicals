```python
import matplotlib.pyplot as plt

def f(x):
    return (x+3)**2

def derivative(x):
    return 2*(x+3)

x = 2
learning_rate = 0.1
iterations = 50

points = [x]

for i in range(iterations):
    x = x - learning_rate * derivative(x)
    points.append(x)

# Plot the function
x_values = [i/10 for i in range(-50, 31)]
y_values = [f(i) for i in x_values]

plt.plot(x_values, y_values, label="y = (x+3)^2")
plt.scatter(points, [f(i) for i in points], color="red", label="Gradient Descent")
plt.scatter(points[0], f(points[0]), color="green", label="Starting Point")
plt.scatter(points[-1], f(points[-1]), color="blue", label="Minimum Point")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Gradient Descent")
plt.legend()
plt.grid()
plt.show()

print("Minimum x:", round(x, 4))
print("Minimum y:", round(f(x), 4))
```
